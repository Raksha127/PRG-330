```python

employee = {
    "rojan": 35000,
    "rista": 25000,
    "rajesh": 15000
}
```

```python
for i in range(5):

    EMPLOYEE_NAME = input("Enter the name of the employee: ").strip().lower()

    if EMPLOYEE_NAME in employee:

        sales = employee[EMPLOYEE_NAME]

        print("Employee found")
        print("Sales:", sales)

        if sales > 30000:
            commission = sales * 0.02
            print("Commission:", commission)

        else:
            print("No commission")

    else:
        print(f"Employee not found named {EMPLOYEE_NAME}")
```
