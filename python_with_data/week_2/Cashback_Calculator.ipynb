```python
minimum_transaction = 1000
cashback_rate = 0.05
maximum_cashback = 200
qr_bonus_amount = 20

print("Cashback rules are ready.")
```

```python
transaction_amount = float(input("Enter transaction amount (Rs.): "))
payment_method = input("Payment method (QR/Other): ").strip().lower()
```

```python

if transaction_amount <= 0:
    print("Invalid transaction amount. Please enter an amount greater than zero.")


elif transaction_amount < minimum_transaction:
    eligible = False
    cashback_after_cap = 0
    qr_bonus = 0
    final_cashback = 0

    print("\n--- CASHBACK SUMMARY ---")
    print(f"Transaction amount: Rs. {transaction_amount:,.2f}")
    print("Eligible for cashback: No")
    print("Reason: The minimum transaction amount is Rs. 1,000.")
    print("Cashback after cap: Rs. 0.00")
    print("QR bonus added: No")
    print("Final total cashback: Rs. 0.00")

else:
    eligible = True


    calculated_cashback = transaction_amount * cashback_rate


    cashback_after_cap = min(calculated_cashback, maximum_cashback)


    if payment_method == "qr":
        qr_bonus = qr_bonus_amount
        qr_bonus_added = "Yes"
    else:
        qr_bonus = 0
        qr_bonus_added = "No"

    final_cashback = cashback_after_cap + qr_bonus

    print("\n--- CASHBACK SUMMARY ---")
    print(f"Transaction amount: Rs. {transaction_amount:,.2f}")
    print("Eligible for cashback: Yes")
    print(f"Calculated 5% cashback: Rs. {calculated_cashback:,.2f}")
    print(f"Cashback after cap: Rs. {cashback_after_cap:,.2f}")
    print(f"QR bonus added: {qr_bonus_added}")
    print(f"QR bonus amount: Rs. {qr_bonus:,.2f}")
    print(f"Final total cashback: Rs. {final_cashback:,.2f}")
```
